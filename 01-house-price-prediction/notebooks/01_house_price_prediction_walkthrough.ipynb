## 1. Data Ingestion & Summary Statistics

```python
df = load_housing_data()
print(f'Dataset Shape: {df.shape}')
df.head()
```

## 2. Data Cleaning & Feature Engineering

```python
cleaned_df, report = clean_housing_data(df)
print('Cleaning Report:', report)
treated_df, outlier_rep = detect_and_handle_outliers(cleaned_df, iqr_multiplier=2.5)
print('Outliers Handled:', outlier_rep)
```

## 3. Train-Test Split & Pipeline Training

```python
X_train, X_test, y_train, y_test = prepare_train_test_data(treated_df, test_size=0.2, random_state=42)
models = train_regression_models(X_train, y_train, random_state=42)
summary_df, predictions = evaluate_models(models, X_test, y_test)
summary_df
```

## 4. Production Inference with HousePricePredictor

```python
predictor = HousePricePredictor()
sample = {
    'bedrooms': 4,
    'bathrooms': 3.0,
    'sqft_living': 2800,
    'sqft_lot': 8000,
    'floors': 2.0,
    'waterfront': 1,
    'view_quality': 3,
    'condition': 4,
    'grade': 9,
    'yr_built': 2012,
    'yr_renovated': 0,
    'neighborhood': 'Lakeside',
}
result = predictor.predict_single(sample)
print(f"Estimated Price: ${result['estimated_price']:,.2f}")
print(f"Valuation Range: ${result['estimated_range_lower']:,.2f} - ${result['estimated_range_upper']:,.2f}")
```
